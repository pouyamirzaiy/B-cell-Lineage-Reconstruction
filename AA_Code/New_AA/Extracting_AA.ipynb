````python
import pandas as pd
import os

input_file = r"C:\Users\LEGION\Downloads\modify\modified\New_AA\vquest_airr.tsv"
output_dir = r"C:\Users\LEGION\Downloads\modify\modified\New_AA\AA_FASTA"

os.makedirs(output_dir, exist_ok=True)

df = pd.read_csv(input_file, sep="\t", low_memory=False)

df.columns = df.columns.str.strip()

seq_col = "sequence_id"
aa_col = "sequence_alignment_aa"

groups = {}

for _, row in df.iterrows():

    seq_id = str(row[seq_col])
    seq = str(row[aa_col])

    if seq.lower() == "nan":
        continue

    # remove IMGT gaps
    seq = seq.replace(".", "")

    parts = seq_id.split("_", 1)
    if len(parts) != 2:
        continue

    dataset = parts[0]
    header = parts[1]

    groups.setdefault(dataset, []).append((header, seq))

for dataset, items in groups.items():

    out_file = os.path.join(output_dir, f"LLC_{dataset}_1_200_AA.fa")

    with open(out_file, "w") as f:
        for h, s in items:
            f.write(f">{h}\n{s}\n")

    print(f"Created {out_file} ({len(items)})")

print("\nDONE")
````

# Notes: How we created AA FASTA files from IMGT output

## 1. Starting data

We started with the IMGT V-QUEST AIRR output file:

- `vquest_airr.tsv`

This file contains sequence IDs (e.g. `dataset01_seq2958@19`), nucleotide sequences, and multiple annotation fields including amino acid-related columns.

We also had two additional IMGT files:

- `4_IMGT-gapped-AA-sequences.txt`
- `5_AA-sequences.txt`

but we did not use them because the TSV file already contained all the structured information we needed.

---

## 2. Important observation about amino acid columns

We checked the column `sequence_aa`, but it was completely empty (all values were NaN).

Because of that, we could not use it.

Instead, we used:

- `sequence_alignment_aa`

This column contains the amino acid sequences produced by IMGT alignment.

---
