```python
import os
import re

input_folder = r"C:\Users\LEGION\Downloads\modify\modified\LLC"
output_file = r"C:\Users\LEGION\Downloads\modify\modified\LLC\ALL_LLC_concatenated.fa"

with open(output_file, "w") as outfile:
    
    for filename in os.listdir(input_folder):
        if filename.endswith(".fa"):
            
            # Extract dataset number
            match = re.search(r'dataset(\d+)', filename)
            if match:
                dataset_number = match.group(1)
            else:
                continue
            
            filepath = os.path.join(input_folder, filename)
            
            with open(filepath, "r") as infile:
                for line in infile:
                    if line.startswith(">"):
                        # Modify header
                        new_header = f">dataset{dataset_number}_{line[1:].strip()}\n"
                        outfile.write(new_header)
                    else:
                        outfile.write(line)

print("Done! All files concatenated.")
```

```python
import os
import re

input_file = r"C:\Users\LEGION\Downloads\modify\modified\LLC\ALL_LLC_concatenated.fa"
output_file = r"C:\Users\LEGION\Downloads\modify\modified\LLC\ALL_LLC_clean.fa"

with open(input_file, "r") as infile, open(output_file, "w") as outfile:
    for line in infile:
        if line.startswith(">"):
            outfile.write(line)
        else:
            cleaned = line.replace("-", "")
            outfile.write(cleaned)

print("Done! Gaps removed.")
```
