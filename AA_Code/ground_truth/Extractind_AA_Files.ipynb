```python
import os
import pandas as pd

base_dir = r"C:\Users\LEGION\Downloads\ground_truth"

for i in range(1, 5):
    folder_name = f"asestra_eco_{i}"
    folder = os.path.join(base_dir, folder_name)

    airr_file = os.path.join(folder, "vquest_airr.tsv")
    fasta_file = os.path.join(folder, "AA_sequences.fasta")

    print(f"Processing folder {folder_name}...")

    # Read AIRR TSV
    df = pd.read_csv(airr_file, sep="\t", low_memory=False)

    # Keep only rows with amino acid sequences
    df = df[df["sequence_alignment_aa"].notna()]

    with open(fasta_file, "w") as f:
        for _, row in df.iterrows():
            seq_id = str(row["sequence_id"]).strip()
            aa_seq = str(row["sequence_alignment_aa"]).replace(" ", "").strip()

            if aa_seq == "" or aa_seq.lower() == "nan":
                continue

            f.write(f">{seq_id}\n")
            f.write(f"{aa_seq}\n")

    print(f"Saved {fasta_file} ({len(df)} sequences)")
```

Generating Amino Acid FASTA Files from IMGT AIRR Output

In this step, we generated amino acid FASTA files for each of the six datasets using the corresponding vquest_airr.tsv files produced by IMGT/V-QUEST.

For each folder (1–4), the script loaded the vquest_airr.tsv file into a pandas DataFrame. We used the sequence_id column as the FASTA header and the sequence_alignment_aa column as the amino acid sequence because the sequence_aa column was empty, whereas sequence_alignment_aa contained the aligned amino acid sequences generated by IMGT.

Before writing the output, the script removed any rows with missing amino acid sequences (NaN) to ensure that only valid sequences were included. It then wrote each remaining sequence to a FASTA file named AA_sequences.fasta, using the standard FASTA format:

>sequence_id
amino_acid_sequence

A separate FASTA file was created in each of the six folders, preserving the organization of the original datasets and producing files that are ready for downstream analyses such as clonal lineage reconstruction, sequence alignment, or phylogenetic inference.
